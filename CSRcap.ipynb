## 5. Clean the data

### 5.1 Year: keeping only the year number
Python string slicing `[:4]` in a list comprehension takes the first 4 characters. For converting whole date to year.

```python
df["year"] = [int(str(v)[:4]) for v in df["year"]]
print(df["year"].value_counts())
```

### 5.2 Converting into small column names (add a new column, delete the old one)


```python
df["outlay_lakhs"] = df["project_outlay_amount_in_lakhs"]
df["spent_lakhs"] = df["amount_spent_in_lakhs"]
df = df.drop(columns=["project_outlay_amount_in_lakhs", "amount_spent_in_lakhs"])
print(list(df.columns))
```

### 5.3 Missing project names (`fillna`)

```python
df["csr_project_name"] = df["csr_project_name"].fillna("Not Mentioned")
print("Missing project names now:", df["csr_project_name"].isnull().sum())
```

### 5.4 Negative amounts
Money spent or planned cannot be negative. So first marking them as missing with `np.nan` (using `loc`), then remove those rows with `dropna`.

```python
print("Negative rows found:", (df["spent_lakhs"] < 0).sum() + (df["outlay_lakhs"] < 0).sum())

df.loc[df["spent_lakhs"] < 0, "spent_lakhs"] = np.nan
df.loc[df["outlay_lakhs"] < 0, "outlay_lakhs"] = np.nan

rows_before = len(df)
df = df.dropna(subset=["spent_lakhs", "outlay_lakhs"])
print("Rows removed:", rows_before - len(df))
```

### 5.5 Impossible outlay values
An outlay above 10,00,000 lakhs (10,000 crore) for one project is a data entry error. Only keeping rows at or below this limit (boolean indexing).

```python
rows_before = len(df)
df = df[df["outlay_lakhs"] <= 1000000]
print("Rows removed:", rows_before - len(df))
```

### 5.6 Inconsistent label: PLC means Public (update using `loc`)

```python
print("Before:")
print(df["company_class"].value_counts())

df.loc[df["company_class"] == "PLC", "company_class"] = "Public"

print()
print("After:")
print(df["company_class"].value_counts())
```

### 5.7 Remove white spaces

Extra spaces make the same text look like different values. The loop below removes spaces at the start, at the end and double spaces inside the text, in all text columns. The missing project names were already filled in step 5.3, so there are no empty values here.



```python
text_cols = ["state_name", "district_name", "csr_project_name", "development_sector",
             "company_name", "cin", "company_class", "company_type", "roc",
             "company_category", "listing_status"]

for col in text_cols:
    df[col] = [" ".join(str(v).split()) for v in df[col]]

print("Unique project names after:", df["csr_project_name"].nunique())
```

### 5.8 Remove duplicates
`drop_duplicates()` keeps the first copy and removes the rest. Compare all columns except `id`.

```python
rows_before = len(df)
df = df.drop_duplicates(subset=[c for c in df.columns if c != "id"])
print("Duplicate rows removed:", rows_before - len(df))
```

### 5.9 Standardise category labels and expand short forms
- `Govt.` is a short form, so writing it in full
- `Prime Minister'S` has a wrong capital S, so correcting it

```python
df.loc[df["company_type"] == "Govt. Companies", "company_type"] = "Government Companies"
df.loc[df["company_type"] == "Other than Govt. Companies", "company_type"] = "Other than Government Companies"
df.loc[df["development_sector"] == "Prime Minister'S National Relief Fund", "development_sector"] = "Prime Minister's National Relief Fund"

print(df["company_type"].value_counts())
print()
print(df["development_sector"].value_counts().tail(8))
```

### 5.10 Adding a new column: spent in crores
1 crore = 100 lakhs. Amounts in crores are easier to read on adding up the totals.

```python
df["spent_crore"] = df["spent_lakhs"] / 100
df[["spent_lakhs", "spent_crore"]].head()
```
