### 4.6 Placeholder values (hidden missing data)

### 4.7 Spent compared with planned outlay
Step 4.4 showed that many projects spent more than planned. Splitting these rows into two types, and checking how big the difference is.

```python
print("Spent more than outlay:", (spent > outlay).sum())
print("   outlay is 0 (planned amount was not recorded):", ((spent > outlay) & (outlay == 0)).sum())
print("   outlay is more than 0 (spent really exceeded the plan):", ((spent > outlay) & (outlay > 0)).sum())

print()
print("Spent more than 2 times the outlay:", ((spent > 2 * outlay) & (outlay > 0)).sum())
print("Spent more than 10 times the outlay:", ((spent > 10 * outlay) & (outlay > 0)).sum())
print("Spent more than 100 times the outlay:", ((spent > 100 * outlay) & (outlay > 0)).sum())
```

- `outlay` and `spent` are the short names created in 4.4.
- `spent > outlay` is True where the project spent more than its plan, and `.sum()` counts those rows.
- `&` means AND, so both conditions must be true. `(spent > outlay) & (outlay == 0)` counts rows that spent money but have a planned amount of 0.
- `(spent > outlay) & (outlay > 0)` counts rows that have a real plan and still spent more.
- `spent > 2 * outlay` checks whether the spending is more than double the plan. The same idea is used for 10 times and 100 times.
- `(outlay > 0)` is added so that rows with a zero plan are not counted again.

**Finding:** Of the 4,505 projects where spent is more than the outlay, 1,790 have an outlay of 0, which means the plan was not recorded and not that money was overspent. The other 2,714 really spent more than planned. Only 353 spent more than 2 times the plan, 61 more than 10 times and 8 more than 100 times. (The 1 remaining row of the 4,505 is the negative outlay that is removed in 5.4.)

**Decision:** All these rows are kept. Spending can really go above a plan, and a zero outlay only means the plan is unknown. Because the outlay column is less reliable, `spent` will be the main measure in the analysis.

## 5. Clean the data

### 5.1 Year: keeping only the year number
Python string slicing `[:4]` in a list comprehension takes the first 4 characters. For converting whole date to year.

```python
df["year"] = [int(str(v)[:4]) for v in df["year"]]
print(df["year"].value_counts())
```

A year number such as `2018` is easier to group and plot than a full date when the analysis is yearly.

- `df["year"]` selects the year column on the left side, where the cleaned result will be stored.
- The list comprehension `[ ... for v in df["year"]]` repeats an operation for every value in the column. Each original value is temporarily called `v`.
- `str(v)` converts the value to text.
- `[:4]` takes the first four characters. For a date string beginning with a four-digit year, these characters represent the year.
- `int(...)` converts those four characters into a whole number.
- The result replaces the original `year` column.
- `df["year"].value_counts()` counts how many records belong to each year

### 5.2 Converting into small column names (add a new column, delete the old one) for easier later analysis.


```python
df["outlay_lakhs"] = df["project_outlay_amount_in_lakhs"]
df["spent_lakhs"] = df["amount_spent_in_lakhs"]
df = df.drop(columns=["project_outlay_amount_in_lakhs", "amount_spent_in_lakhs"])
print(list(df.columns))
```

- `df["outlay_lakhs"] = creates a new, shorter column name and copies the values from the original outlay column.
- `df["spent_lakhs"] =  creates a shorter name for the amount-spent column.
- `df.drop(columns= removes the two original long-named columns.
- `df =  saves the DataFrame returned by `drop()` back into `df`.
- `print(list(df.columns))` displays the updated column names so we can confirm the change.

### 5.3 Missing project names (`fillna`)

```python
df["csr_project_name"] = df["csr_project_name"].fillna("Not Mentioned")
print("Missing project names now:", df["csr_project_name"].isnull().sum())
```

Keeping a clear label can make category summaries easier to interpret than leaving a blank project name.

- `df["csr_project_name"]` selects the project-name column.
- `.fillna("Not Mentioned")` replaces recognised missing values in that column with the text `Not Mentioned`.
- The assignment stores the filled result back into the same column.
- The final expression `.isnull().sum()` counts how many missing project names remain.
- `print()` displays that remaining count.

### 5.4 Negative amounts
Negative project outlay or spending is treated as invalid for this analysis. Here first marks those values as missing and then removes records that have a missing value in either amount column.

```python
print("Negative rows found:", (df["spent_lakhs"] < 0).sum() + (df["outlay_lakhs"] < 0).sum())

df.loc[df["spent_lakhs"] < 0, "spent_lakhs"] = np.nan
df.loc[df["outlay_lakhs"] < 0, "outlay_lakhs"] = np.nan

rows_before = len(df)
df = df.dropna(subset=["spent_lakhs", "outlay_lakhs"])
print("Rows removed:", rows_before - len(df))
```

- The first `print()` counts negative values in both money columns and displays the combined count. A row with negative values in both columns contributes 2 to this total because the two column counts are added.
- `df.loc[condition, column]` selects cells using a row condition and a column name.
- `df.loc[df["spent_lakhs"] < 0, "spent_lakhs"] = np.nan` replaces negative spending values with `NaN` (Pandas' missing-value marker).
- The next `loc` line does the same for negative outlay values.
- `rows_before = len(df)` stores the number of rows before removal.
- `df.dropna(subset=["spent_lakhs", "outlay_lakhs"])` removes rows where either of these two selected amount columns is missing.
- The result is assigned back to `df`.
- The final print subtracts the new row count from the old row count to show how many rows were removed.

### 5.5 Impossible outlay values
An outlay above 10,00,000 lakhs (10,000 crore) for one project is a data entry error. Only keeping rows at or below this limit (boolean indexing).

```python
rows_before = len(df)
df = df[df["outlay_lakhs"] <= 1000000]
print("Rows removed:", rows_before - len(df))
```

- `rows_before = len(df)` records how many rows exist before filtering.
- `df["outlay_lakhs"] <= 1000000` checks each row and returns True when outlay is at or below the chosen maximum.
- `df = df[...]` keeps only rows for which that condition is True. Rows above the limit are excluded.
- The final print calculates the difference between the original and filtered row counts.

### 5.6 Inconsistent label: PLC means Public (update using `loc`)

If `PLC` and `Public` are intended to represent the same category in this dataset, standardising the label prevents them from being counted separately.


```python
print("Before:")
print(df["company_class"].value_counts())

df.loc[df["company_class"] == "PLC", "company_class"] = "Public"

print()
print("After:")
print(df["company_class"].value_counts())
```

- The first two `print()` statements show the category counts in `company_class` before the change.
- `df.loc[df["company_class"] == "PLC", "company_class"] = "Public"` finds rows where the company class is exactly `PLC` and changes that cell to `Public`.
- The blank `print()` adds spacing.

### 5.7 Remove white spaces

Extra spaces make the same text look like different values. The loop below removes spaces at the start, at the end and double spaces inside the text, in all text columns.

For example, `Kerala`, ` Kerala `, and `Kerala   ` should not be treated as three different labels just because of extra spaces.



```python
text_cols = ["state_name", "district_name", "csr_project_name", "development_sector",
             "company_name", "cin", "company_class", "company_type", "roc",
             "company_category", "listing_status"]

for col in text_cols:
    df[col] = [" ".join(str(v).split()) for v in df[col]]

print("Unique project names after:", len(df["csr_project_name"].value_counts()))
```

- `text_cols = [...]` creates a Python list containing the names of the columns that hold text or category information.
- `for col in text_cols:` loops through those column names one by one.
- `df[col]` selects the current column.
- The list comprehension processes every value `v` in that column.
- `str(v)` converts the value to text.
- `.split()` separates the text into words using whitespace. It also treats repeated spaces, tabs, and line breaks as separators.
- `" ".join(...)` joins the words back together using one normal space between each word. This removes spaces at the beginning and end and reduces repeated internal whitespace.
- The cleaned list is assigned back to the same column.
- The final print counts distinct project-name values after cleaning.

### 5.8 Remove duplicates
`drop_duplicates()` keeps the first copy and removes the rest. Compare all columns except `id`.

```python
rows_before = len(df)
df = df.drop_duplicates(subset=[c for c in df.columns if c != "id"])
print("Duplicate rows removed:", rows_before - len(df))
```

- `rows_before = len(df)` stores the number of rows before duplicate removal.
- `df.columns` provides all column names.
- `[c for c in df.columns if c != "id"]` builds a list of every column except `id`. This is a list comprehension with a condition: include a column name only when it is not `id`.
- `df.drop_duplicates(subset=[])` removes later rows that match an earlier row across all selected columns.
- By default, `drop_duplicates()` keeps the first matching row.
- The result is assigned back to `df`.
- The final print compares the row counts to show how many rows were removed

### 5.9 Standardise category labels and expand short forms

Different spellings or abbreviations can split one real category into multiple groups in charts and summaries. Standardising labels makes grouping more consistent.

- `Govt.` is a short form, so writing it in full
- `Prime Minister'S` has a wrong capital S, so correcting it

```python
df.loc[df["company_type"] == "Govt. Companies", "company_type"] = "Government Companies"
df.loc[df["company_type"] == "Other than Govt. Companies", "company_type"] = "Other than Government Companies"
df.loc[df["development_sector"] == "Prime Minister'S National Relief Fund", "development_sector"] = "Prime Minister's National Relief Fund"

print(df["company_type"].value_counts())
print()
print(df["development_sector"].value_counts().tail(8))
```

- Each `df.loc[condition, column] = new_value` statement updates a selected category label.
- The first line changes `Govt. Companies` to `Government Companies` in `company_type`.
- The second line changes `Other than Govt. Companies` to `Other than Government Companies`.
- The third line corrects the capitalisation in `Prime Minister'S National Relief Fund` to `Prime Minister's National Relief Fund` in `development_sector`.
- The final print statements show the updated counts for company types and the last eight development-sector categories.
